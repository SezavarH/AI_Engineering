### OOP for AI Engineers

- Why use OOP? 

    - Say you need stream, generate, and embed. 
    - As separate functions, you'd have to pass all the data to each one. 
    - As methods of a class, they share the same data and work together automatically—cleaner and simpler.

- Attribute

    - Each object of a class has data, called attributes (e.g., name, age, ...).

- Method

    - Defines the behavior of the object.

- Self

    - A reference to the object itself. It allows us to access the object's attributes and methods.

- __init__

    - The constructor. It runs automatically when a new object is created.

```python
class Car:

    def __init__(self, name, country):
        self.name = name
        self.country  = country

    def show_info(self):
        return f"{self.name} is assembled in {self.country}"

# object
C = Car("Sportage", "Korea")
C.show_info()
```

### Encapsulation

- Meaning?

    Bundling attributes and methods together within a class.
    Controlling access levels: Public, Protected, and Private.
    
- Public

    Default in Python. All attributes and methods are public unless specified otherwise, meaning they can be accessed from anywhere.

- Protected

    Defined by a single underscore _ before the attribute name (e.g., _name).

    Acts like a sign: "Private office, please don't enter" — you can enter if you want, but you're politely being asked not to. 
    It's a convention, not a strict rule.

- Private

    Defined by a double underscore __ before the attribute name (e.g., __name).

    Cannot be accessed directly from outside the class. 
    
    Python performs name mangling to make it harder to access, but it's still technically possible if you know how.

```python
class BankAccount:

    def __init__(self):
        self.balance = 1000            

account = BankAccount()

account.balance                 
```

```python
class BankAccount:

    def __init__(self):
        self._balance = 1000            # "Internal use only"

account = BankAccount()

account._balance                 
```

```python
class BankAccount:

    def __init__(self):
        self.__balance = 1000            # Private

account = BankAccount()

account._balance                        # will raise an error: AttributeError
```

### Inheritance & Polymorphism

- Meaning

    - Inheritance: 
        - A child class inherits attributes and methods from a parent class. (Structure sharing)

    - Polymorphism: 
        - Implemented using inheritance, 
        - but when child classes override (modify/change) the parent's methods to have their own behavior, it's called polymorphism. 
        - Same method name, different behavior

```python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):        # override the behavior
        return "Woof!"

class Cat(Animal):
    def speak(self):        # override the behavior
        return "Meow!"
```

#### super()

Call parent implementation.

```python
class Animal:

    def __init__(self, name):
        self.name = name

class Dog(Animal):

    def __init__(self, name):
        super().__init__(name)

D = Dog("Alex")
D.name
```

```python
class Parent:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print("Parent __init__ called")

class ChildWithoutSuper(Parent):
    def __init__(self, name, age, grade):
        self.grade = grade
        print("Child __init__ called")

# تست:
c1 = ChildWithoutSuper("Ali", 20, "A")
print(c1.grade)   # ✅ وجود داره
print(c1.name)    # ❌ AttributeError! چون name تعریف نشده
print(c1.age)     # ❌ AttributeError! چون age تعریف نشده
```

### Composition (VERY IMPORTANT)

- What is it?

    - A class, instead of inheriting from another class, uses instances of other classes as its attributes (members).

    - In other words: "has-a" relationship, rather than "is-a" relationship.

    - Example: A Car class has an Engine object as a member, instead of inheriting from Engine

```python
## Inheritance

class EmbeddingModel:
    pass

class RAG(EmbeddingModel):
    pass

# RAG IS-A EmbeddingModel 
```

```python
## Composition

class EmbeddingModel:
    pass

class RAG:
    def __init__(self):
        self.embedding = EmbeddingModel()       # instead of inhertance, put an instance

# RAG HAS-A EmbeddingModel
```

```python
# Car and Engine
class Engine:
    
    def start(self):
        print("Engine started ...")

class Car:

    def __init__(self):
        self.engine = Engine()

bmw = Car()
bmw.engine.start()
```

#### @staticmethod, @classmethod

- staticmethod:

    - Used when a method doesn't need to access attributes (neither class nor instance).

- classmethod

    - Used when a method needs to access class attributes (via `cls`) and has no need for instances.

```python
class Car:
    total_cars = 0                                          # class attribute

    def __init__(self, brand):
        self.brand = brand
        Car.total_cars += 1

    @classmethod
    def show_total_cars(cls):                               # access to the class attribute, no need of instances
        return f"Total cars: {cls.total_cars}"

    @staticmethod
    def is_used(year):
        return year >= 1
    
print(Car.show_total_cars())  # Total cars: 0
car1 = Car("BMW")
car2 = Car("Toyota")
print(Car.show_total_cars())  # Total cars: 2
print(Car.is_used(2))     
    
```
