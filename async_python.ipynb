### Async Programming

#### Asyncio

- Core Facts

    - Single Thread & Single Process
    - Concurrency (NOT Parallelism)
    - One Event Loop + One or More Coroutines

- Event Loop

    - Manager that handles concurrency
    - Switches between coroutines
    - Typically one per thread

- Coroutine

    - Function defined with async def
    - Returns a coroutine object (doesn't run immediately)
    - Can pause and resume
    - Won't execute without await or create_task

- await

    - Pauses the coroutine
    - Returns control to Event Loop
    - Resumes when result is ready
    - Can only be used inside async functions

- create_task

    - Schedules coroutine to run in the background
    - Returns a Task object
    - Doesn't block immediately
    - Task runs in background, but need "await" to get the result

- gather

    - Runs multiple coroutines concurrently
    - Starts them at the same time
    - Returns list of results
    - Usually used with await

```python
import asyncio

async def worker(name, delay):
    print(f"{name}: start...")

    await asyncio.sleep(delay)                  # pause the co-routine

    print(f"{name}: finish...")

    return f"{name} with {delay}"

async def main():

    results = []

    # method 1
    task1 = asyncio.create_task(worker("A", 2))                 # run in background
    task2 = asyncio.create_task(worker("B", 1))

    results.append(await task1)                                 # get the result
    results.append(await task2)
    print(results)

    # method 2
    results_g = await asyncio.gather(                           # run concurrently
        worker("C", 3),
        worker("D", 1)
    )
    print(results_g)

await(main())
```

### Thread

- One or more threads exist inside each process

- Threads share the same memory space (heap)

- GIL (Global Interpreter Lock): Only one thread can execute Python bytecode at a time

- Multi-threading is efficient for I/O-bound tasks (but not CPU-bound)

- Context Switching:	Threads are lightweight, switching is faster than processes

```python
import threading
import time

def io_task():
    time.sleep(1)  # I/O operation (releases GIL)
    print("Done")

# Efficient for I/O
threads = [threading.Thread(target=io_task) for _ in range(10)]
for t in threads:
    t.start()
for t in threads:
    t.join()
```
