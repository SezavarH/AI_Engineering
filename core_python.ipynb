### Variable

- variable

    - a name binding not a storage space
    - x = 10 ==> x -----> |object: int = 10|

- object

    - every thing is object in python
    - mutable or immutable



```python
x = 10
y = x

x += 1
print(y)
print(x)                                            # Reason: int is immutable, it creates a new object for x but y still references to the 10
```

```python
x = [1, 2]
y = x

x.append(3)
print(y)
print(x)                                            # Reason: list is mutable, just change the value of object 
```

```python
x = "hello"
y = x

x += " world"

print(x)
print(y)                                            # Reason: string is immutable
```

### Functions

- First class object

    - function is a first class object, means it can store in a variable, and pass to another function 

```python
f = lambda x: x+2

# even function is an object (callable object that encapsulate a behaviour)
```

```python
# function with mutable argument
def count(tmp = []):

    tmp.append("*")

    return tmp

count()
count()
count()                                         # only one time [] is created 
```

```python
# function with immutable argument
def count(tmp = ""):

    tmp = tmp.join("*")

    return tmp

count()
count()
count()                                        
```

### Iterator & Generator

- Iterator

    - can use for loop over it

- Generator

    - an iterator that has "yield"
    - yield can pause a function: enable streaming process

```python
# list as a iterator
my_list = [1, 2, 3]
```

```python
def count():
    yield 1
    yield 2
    yield 3

g = count()
next(g)
```

```python
next(g)
```

```python
# yield from
def inner_count():
    yield 2
    yield 3

def count():
    yield 1
    yield from inner_count()
    yield 4

g = count()
next(g)
```

```python
next(g)
```

```python
next(g)
```

### Decorators

change bahavior of a function

```python
import time

def run_time(func):                                 # first pass the func
    def wrapper(n):                                 # then pass the arg to the wrapper

        t1 = time.time()
        print("Started ...")
        result = func(n)
        print(f"Finished in {time.time()- t1}")


        return result

    return wrapper                                  # return wrapper at the end (not func)

@run_time
def create(n):
    tmp = range(n)

create(100000)
```

### Type hint

```python
# input should be str or None or int

from typing import Optional

def predict(var: str | int | None) -> int:             
    print(var)

predict(3)
predict("ali")
predict(None)
```

### TypedDict

a dictionary that its keys are known

```python
from typing import TypedDict

class State(TypedDict):

    lr : float
    epoch: int

S : State = {"lr": 1.12, "epoch": 5}
S
```

```python
class State(TypedDict):

    question: str
    answer: str

S: State = {
    "question" : "hi there",
    "answer"   : "hi",
    "any": 12                               # doesn't raise an error
}
S
```
